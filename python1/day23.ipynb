# Decorators

```python
def greet():
    print("Hello, World!")

x = greet
x()
```

```python
def greet():
    print("Hello")

def call_function(func):
    func()

call_function(greet)
```

```python
def outer():
    def inner():
        print("Inner function")
    return inner()

outer()
```

```python
def outer():
    def inner():
        print("Inner function")
    return inner

f = outer()
f()
```

## What is Decorators
#### When a function takes another function wraps it and return a new function

```python
def decorator(func):
    def wrapper():
        print("Before function call")
        func()
        print("After function call")
    return wrapper
```
