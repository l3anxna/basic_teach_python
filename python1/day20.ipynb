## Boolean Masking (Most powerful)

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[arr > 25])
```

```python
arr[arr % 20 == 0]
```

```python
arr[arr > 25] = 0
print(arr)
```

## Broadcasting

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print(arr + 10)
```

### Broadcasting Rules
- Rule 1: If dimensions are diiferent, Numpy compares from right to left
- Rule 2: Two dimensions are compatible if: they are equal or one of them is 1

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]]) #shape: (2, 3)

B = np.array([10, 20, 30]) #shape: (1, 3)

print(A + B)
```
