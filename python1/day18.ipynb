## Numpy Arrays vs Python Lists

```python
# Python Lists
my_list = [1, 2, 3, 4, 5]

# Numpy Arrays
my_array = np.array([1, 2, 3, 4, 5])

print("Python List:", my_list)
print("Numpy Array:", my_array)
```

```python
print(my_array * 2)
```

```python
print(my_list * 2)
```

## Array shape and Reshape

### Shape = Structure of array

it tells us
- how many rows
- how many columns

```python
arr = np.array([1, 2, 3, 4, 5])
print(arr.shape)
```

```python
arr2 = np.array([[1, 2, 3], 
                 [4, 5, 6]])
print(arr2.shape)
```

```python
arr3 = np.zeros((2, 3, 4))
print(arr3.shape)
```

### What is Reshape
Reshape = Chnage structure without changing data

```python
arr = np.arange(6)
print(arr)
```

```python
arr2 = arr.reshape((2, 3))
print(arr2)
```

```python
arr = np.arange(12)
print(arr)
```

```python
arr2 = arr.reshape((3, -1))
print(arr2)
```

```python
arr.ndim
arr.size
arr.shape
arr.reshape()
```

```python
arr = np.arange(8).reshape((2, 4))

print(arr.ndim)
print(arr.size)
print(arr.shape)
```
