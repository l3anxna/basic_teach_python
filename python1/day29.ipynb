### Iterable VS Iterator

#### What is Iterable

An object capable of returning its member one at a time

- list
- tuple
- dictionary
- set

#### What is Iterator

An object that remembers its state during iterations

Must implement:
- \__iter__
- \__next__

```python
nums = [10, 20, 30]
it = iter(nums)

print(next(it))
print(next(it))
print(next(it))
print(next(it))
```

```python
it = iter(nums)

for item in nums:
    while True:
        try:
            item = next(it)
            print(item)
        except StopIteration:
            break
```

```python
class Counter:
    def __init__ (self, max_value):
        self.max_value = max_value
        self.current = 0

    def __iter__(self):
        return self
    
    def __next__(self):
        if self.current <= self.max_value:
            value = self.current
            self.current += 1
            return value
        else:
            raise StopIteration
```

```python
x = Counter(5)

for num in x:
    print(num)
```
