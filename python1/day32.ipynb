```python
#Exercise
import threading
import time

files = [
    "lecture1.pdf",
    "lecture2.pdf",
    "assignment.zip",
    "slides.pptx",
    "python_notes.pdf"
]

def download(file_name):
    print(f"Downloading {file_name}...")
    time.sleep(2)
    print(f"Finished {file_name}!")

threads = []

for file in files:
    thread = threading.Thread(target=download, args=(file,))
    threads.append(thread)
    thread.start()

for thread in threads:
    thread.join()

print("All downloads completed!")
```

```python
import time
from multiprocessing import Process
start = 0
end = 50000000

def calculate(start, end):
    total = 0
    for i in range(start,end):
        total += i*i
    
    print(f"Total {total}")


if __name__ == "__main__":

    start_time = time.time()

    ranges = [(0, 12500000), (12500000, 25000000), (25000000, 37500000), (37500000, 50000000)]

    processes = []

    for start, end in ranges:
        p = Process(target=calculate, args=(start,end))
        processes.append(p)
        p.start()

    for p in processes:
        p.join()

    end_time = time.time()

    print(f"Time: {end_time - start_time:.2f} seconds")
```
