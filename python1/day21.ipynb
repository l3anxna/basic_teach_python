# Matplotlib

## Data Visualisation

- Understanding trend
- Comparing categories
- Detecing anomalies

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]

plt.plot(x, y)

plt.title("Sales Growth")
plt.xlabel("Year")
plt.ylabel("Revenue")

plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr"]
product_a = [100, 150, 200, 250]
product_b = [80, 120, 160, 220]

plt.plot(months, product_a)
plt.plot(months, product_b)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend(["Product A", "Product B"])

plt.show()
```

```python
products = ["Laptop", "Smartphone", "Tablet", "Headphones"]
sales = [300, 500, 200, 400]

plt.bar(products, sales)

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

```python
import numpy as np

data = np.random.randn(1000)

plt.hist(data, bins=(30))

plt.title("Histogram of Random Data")

plt.show()
```

```python
x = np.random.rand(100)
y = np.random.rand(100)

plt.scatter(x, y)

plt.title("Scatter Plot")

plt.show()
```

```python
fig, axs = plt.subplots(1, 4, figsize=(20, 5))

axs[0].plot([1, 2, 3], [10, 20, 30])
axs[0].set_title("Line Plot")

axs[1].bar(["A", "B", "C"], [5, 10, 15])
axs[1].set_title("Bar Chart")

plt.show()
```

```python
plt.plot([1, 2, 3], [10, 20, 30])
plt.title("Line Plot with grid")

plt.grid(True)
plt.show()
```
