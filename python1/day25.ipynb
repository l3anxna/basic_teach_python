# Decorator with Arguments

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):  # universal adapter
        print("Before")
        func(*args, **kwargs)
        print("After")
    return wrapper

@my_decorator
def add(x, y):
    print(x + y)

add(10, 20)
```

```python
def func(*args):
    print(args)
```

```python
func(1, 2, 3)
```

```python
def func2(**kwargs):
    print(kwargs)
```

```python
func2(name="Alice", age=30, city="New York", country="USA")
```

```python
def func3(*args, **kwargs):
    print("Positional arguments:", args)
    print("Keyword arguments:", kwargs)
func3(1, 2, 3, name="Alice", age=30)
```

```python
@debug
def add(x, y):
    return x + y

add(2, 3)
```

```python
def debug(func):
    def wrapper(*args, **kwargs):
        print(f"Function called with: {args} {kwargs}")
        return func(*args, **kwargs)
    return wrapper

@debug
def greet(name):
    print("Hello", name)

greet("Arm")
```

```python
def debug(func):
    def wrapper(*args, **kwargs):
        print(f"the function is called with args={args}, kwargs={kwargs}")
        return func(*args, **kwargs)
    return wrapper
@debug
def add(a,b):
    print(a+b)
add(1,2)
```
