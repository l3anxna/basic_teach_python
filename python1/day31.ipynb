# Threading

```python
import threading

def task():
    print("Working...")

t = threading.Thread(target=task)

t.start()
t.join()

print("Work completed.")
```

```python
import time

def download():
    time.sleep(3)

download()
download()
download()
```

```python
import threading
import time

def download():
    time.sleep(3)

threads = []

for _ in range(3):
    t = threading.Thread(target=download)
    threads.append(t)
    t.start()

for t in threads:
    t.join()
```

## Race Condition

- Multiple threads access shared data and result depends on timing

```python
counter = 0
```

```python
lock = threading.Lock()

with lock:
    counter += 1
```
