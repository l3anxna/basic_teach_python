# Multiple Decorators (Stacking)

```python
def A(func):
    def wrapper():
        print("A start")
        func()
        print("A end")
    return wrapper

def B(func):
    def wrapper():
        print("B start")
        func()
        print("B end")
    return wrapper
```

```python
@A
@B
def greet():
    print("Hello")

greet()
```

```python
def star(func):
    def wrapper():
        print("*****")
        func()
        print("*****")
    return wrapper

def hash(func):
    def wrapper():
        print("#####")
        func()
        print("#####")
    return wrapper
```

```python
@hash
@star
def show():
    print("Welcome")

show()
```

```python
from functools import wraps
def my_decorator(func):
    @wraps(func)
    def wrapper():
        print("Before")
        func()
    return wrapper
```

```python
@my_decorator
def say_hi():
    print("Hi")

print(say_hi.__name__)
```

```python
from functools import wraps

logged_in = True

def logger(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Calling function: {func.__name__}")
        return func(*args, **kwargs)
    return wrapper


def border(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print("========")
        result = func(*args, **kwargs)
        print("========")
        return result
    return wrapper


def authenticate(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        if logged_in:
            return func(*args, **kwargs)
        else:
            print("Access denied")
    return wrapper
```

```python
@authenticate
@logger
@border
def secret_page():
    print("Welcome")

secret_page()

print(secret_page.__name__)
```
