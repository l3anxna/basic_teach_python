# Decorators

```python
def say_hi():
    print("Hi!")

say_hi()
```

```python
def decorator(func):
    def wrapper():
        print("BEFORE")
        func() # Call the original function (say_hi)
        print("AFTER")
    return wrapper
```

```python
say_hi = decorator(say_hi)
say_hi()
```

```python
@decorator
def say_hi():
    print("Hi!")
```
