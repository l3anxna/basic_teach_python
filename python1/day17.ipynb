# Project Management in Python

## Conda

#### Environment and package manager

- Manages Python version
- Manages environment
- Manages non-python libraries


### Both of Conda and Venv create isolated environment 
### Difference is what they can manage and how much control they give you

#### Venv

##### What it is
- Python-only virtual environment
- Comes with python
- uses pip install package

##### What it can manage
- Python packages

#### Conda

##### What it is
- Environment and Packages manage
- Not limites to Python packages
- Mange Python version per environment

##### What it can manage
- Python packages
- Python Version
- Non-Python dependencies (C, C++, CUDA, etc.)
