```python
def repeat(n):   # receives parameter
    def decorator(func):  # receives function
        def wrapper(*args, **kwargs):  # runs function
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hi():
    print("hi")

say_hi()
```

```python
def limit(n):
    def decorator(func):
        count = 0
        def wrapper(*args,**kwargs):
            nonlocal count
            if count < n:
                count += 1
                return func(*args,**kwargs)
            else:
                print("reached limit")
        return wrapper
    return decorator
```

```python
@limit(3)
def sum1(a,b):
    c = a + b
    print(c)
```

```python
sum1(2,3)
sum1(3,4)
sum1(4,5)
result = sum1(5,6)
```
