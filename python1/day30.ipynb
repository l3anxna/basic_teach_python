## What is a Generator?

### A simpler way to create iterator

Instead of:
- implementing
- writing __iter__
- writing __next__

Use: 
- yield

```python
def count_up_to(max_num):
    num = 1
    while num <= max_num:
        yield num
        num += 1
```

```python
counter = count_up_to(10)

for n in counter:
    print(n)
```

### Yield

1. Pause function
2. Saves state
3. Resumes later

```python
def test():
    print("A")
    yield 1
    print("B")
    yield 2
    print("C")
```

```python
t = test()
print(next(t))
print(next(t))
print(next(t))
```

```python
nums = [x*x for x in range(5)]
```

```python
nums = (x*x for x in range(5))
```

```python
import sys

list_nums = [x*x for x in range(10000000)]
gen_nums = (x*x for x in range(10000000))

print(sys.getsizeof(list_nums))
print(sys.getsizeof(gen_nums))
```

### Infinite Generators

```python
def infinite():
    num = 1
    while True:
        yield num
        num += 1
```

```python
gen = infinite()

for _ in range(10):
    print(next(gen))
```

### yield from

```python
def gen1():
    yield 1
    yield 2

def gen2():
    yield from gen1()
    yield 3
```

```python
gen2_instance = gen2()
print(next(gen2_instance))
print(next(gen2_instance))
print(next(gen2_instance))
```

```python
def echo():
    while True:
        value = yield
        print(f"Received: {value}")
```

```python
e = echo()
next(e)

e.send("Hello")
e.close() 
e.throw(StopIteration)
```
