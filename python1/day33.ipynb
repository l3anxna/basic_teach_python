### Event Loop

```python
async def fetch(n):
    await asyncio.sleep(n)
```

```python
await fetch(1)
await fetch(2)
await fetch(3)
```

```python
await asyncio.gather(fetch(1), fetch(2), fetch(3))
```

#### Thread
- I/O
- Shared Memory
- GIL Affected
- Not good for thousand of tasks
- Complexity : Normal

#### Process
- CPU
- Seperated Memory
- GIL Not Affected
- Not good for thousand of tasks
- Complexity : Normal

#### AsyncIO
- Massive I/O
- Shared Memory
- Single thread
- Good for thousand of tasks
- Complexity : High

### Task 3 — AsyncIO (15 minutes)
Scenario : Now the application needs to fetch user information from three remote APIs.
Create : async def fetch(user_id):
Inside : await asyncio.sleep(2)
Return
{
    "id": user_id,
    "status": "OK"
}
First

Run sequentially.

await fetch(1)
await fetch(2)
await fetch(3)

Measure runtime.

Then

Run concurrently.

await asyncio.gather(
    fetch(1),
    fetch(2),
    fetch(3)
)

Measure runtime again.

#### Questions
- Why is await asyncio.sleep() important?
- What happens if we replace it with time.sleep()?
- What is the Event Loop doing while one coroutine is waiting?
