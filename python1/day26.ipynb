# Returning value from decorator

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Before")
        result = func(*args, **kwargs)
        print("After")
        return result
    return wrapper
```

```python
@my_decorator
def multiply(x, y):
    return x * y

print(multiply(2, 3))
```

```python
@my_decorator
def square(x):
    return x * x

print(square(4))
```

# Real-World Use Case

## Timer Decorator

How do we measure how long a functions takes?

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print("Time: ", end - start)
        return result
    return wrapper
```

```python
@timer
def slow_function():
    for _ in range(1000):
        pass

print(slow_function())
```

```python
def authenticate(func):
    def wrapper(username, password):
        if username == "admin" and password == "jjl":
            return func(username, password)
        else:
            return "Authentication failed"
    return wrapper
```
